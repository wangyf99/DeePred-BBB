```python
true_positive = ((merged_df['BBB-Class'] == 1) & (merged_df['Predicted_class'] == 1)).sum()
true_positive
```

```python
false_positive = ((merged_df['BBB-Class'] == 0) & (merged_df['Predicted_class'] == 1)).sum()
false_positive
```

```python
precision = true_positive / (true_positive + false_positive)
precision
```

```python
false_negative = ((merged_df['BBB-Class'] == 1) & (merged_df['Predicted_class'] == 0)).sum()
false_negative
```

```python
recall = true_positive / (true_positive + false_negative)
recall
```

```python
(2 * precision * recall)/(precision + recall)
```
